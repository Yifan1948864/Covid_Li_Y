```python
# Random Forest models
%run 07_model_functions.py

import json
import os
import optuna

optuna.logging.set_verbosity(optuna.logging.WARNING)
```

```python
# Low policy: face-mask outcome
X_train, y_train, X_test, y_test = load_train_test_data(
    "low_face_mask_train.csv",
    "low_face_mask_test.csv",
    "face_mask_behaviour_binary"
)

rf_low_face_mask_study = tune_random_forest(
    X_train, y_train, n_estimators=500, n_trials=200
)

rf_low_face_mask_result = {
    "best_cv_auc": rf_low_face_mask_study.best_value,
    "best_params": rf_low_face_mask_study.best_params
}
with open(
    "../Model/rf_low_policy_face_mask_params.json", "w"
) as file:
    json.dump(rf_low_face_mask_result, file, indent=2)

rf_low_face_mask = fit_random_forest(
    X_train,
    y_train,
    rf_low_face_mask_study.best_params,
    n_estimators=500
)
rf_low_face_mask_metrics = evaluate_binary_model(
    rf_low_face_mask, X_test, y_test
)

rf_low_face_mask_importance = repeated_feature_importance(
    rf_low_face_mask, X_train, y_train, n_repeats=100
)
rf_low_face_mask_importance.to_csv(
    "../Importance/rf_low_policy_face_mask_importance.csv",
    index=False
)

display(rf_low_face_mask_metrics)
display(rf_low_face_mask_importance.head(10))
```
