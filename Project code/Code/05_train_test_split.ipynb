```python
import pandas as pd
from sklearn.model_selection import train_test_split
```

```python
# Read datasets for the three policy intensity levels
low_policy_data = pd.read_csv("../Cleaned data/low_policy_data.csv")
medium_policy_data = pd.read_csv("../Cleaned data/medium_policy_data.csv")
high_policy_data = pd.read_csv("../Cleaned data/high_policy_data.csv")
```

```python
# Low policy intensity: face-mask outcome
low_face_mask_data = low_policy_data.drop(
    columns=["protective_behaviour_binary"]
)

low_face_mask_train, low_face_mask_test = train_test_split(
    low_face_mask_data,
    test_size=0.2,
    random_state=3034,
    stratify=low_face_mask_data["face_mask_behaviour_binary"]
)


low_face_mask_train.to_csv(
    "../Train and test/low_face_mask_train.csv", index=False
)
low_face_mask_test.to_csv(
    "../Train and test/low_face_mask_test.csv", index=False
)
```

```python
# Low policy intensity: protective-behaviour outcome
low_protective_data = low_policy_data.drop(
    columns=["face_mask_behaviour_binary"]
)

low_protective_train, low_protective_test = train_test_split(
    low_protective_data,
    test_size=0.2,
    random_state=3034,
    stratify=low_protective_data["protective_behaviour_binary"]
)


low_protective_train.to_csv(
    "../Train and test/low_protective_train.csv", index=False
)
low_protective_test.to_csv(
    "../Train and test/low_protective_test.csv", index=False
)
```

```python
# Medium policy intensity: face-mask outcome
medium_face_mask_data = medium_policy_data.drop(
    columns=["protective_behaviour_binary"]
)

medium_face_mask_train, medium_face_mask_test = train_test_split(
    medium_face_mask_data,
    test_size=0.2,
    random_state=3034,
    stratify=medium_face_mask_data["face_mask_behaviour_binary"]
)

medium_face_mask_train.to_csv(
    "../Train and test/medium_face_mask_train.csv", index=False
)
medium_face_mask_test.to_csv(
    "../Train and test/medium_face_mask_test.csv", index=False
)
```

```python
# Medium policy intensity: protective-behaviour outcome
medium_protective_data = medium_policy_data.drop(
    columns=["face_mask_behaviour_binary"]
)

medium_protective_train, medium_protective_test = train_test_split(
    medium_protective_data,
    test_size=0.2,
    random_state=3034,
    stratify=medium_protective_data["protective_behaviour_binary"]
)

medium_protective_train.to_csv(
    "../Train and test/medium_protective_train.csv", index=False
)
medium_protective_test.to_csv(
    "../Train and test/medium_protective_test.csv", index=False
)
```

```python
# High policy intensity: face-mask outcome
high_face_mask_data = high_policy_data.drop(
    columns=["protective_behaviour_binary"]
)

high_face_mask_train, high_face_mask_test = train_test_split(
    high_face_mask_data,
    test_size=0.2,
    random_state=3034,
    stratify=high_face_mask_data["face_mask_behaviour_binary"]
)

high_face_mask_train.to_csv(
    "../Train and test/high_face_mask_train.csv", index=False
)
high_face_mask_test.to_csv(
    "../Train and test/high_face_mask_test.csv", index=False
)
```

```python
# High policy intensity: protective-behaviour outcome
high_protective_data = high_policy_data.drop(
    columns=["face_mask_behaviour_binary"]
)

high_protective_train, high_protective_test = train_test_split(
    high_protective_data,
    test_size=0.2,
    random_state=3034,
    stratify=high_protective_data["protective_behaviour_binary"]
)

high_protective_train.to_csv(
    "../Train and test/high_protective_train.csv", index=False
)
high_protective_test.to_csv(
    "../Train and test/high_protective_test.csv", index=False
)
```
