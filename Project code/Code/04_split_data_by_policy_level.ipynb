```python
import pandas as pd
```

```python
# Read the merged dataset
merged_data = pd.read_csv("../Cleaned data/merged_data.csv")
```

```python
# Split the dataset by policy intensity level
low_policy_data = merged_data[
    merged_data["policy_intensity_level"] == "low"
].copy()

medium_policy_data = merged_data[
    merged_data["policy_intensity_level"] == "medium"
].copy()

high_policy_data = merged_data[
    merged_data["policy_intensity_level"] == "high"
].copy()
```

```python
# Remove the grouping variable after validating the split
low_policy_data = low_policy_data.drop(columns=["policy_intensity_level"])
medium_policy_data = medium_policy_data.drop(columns=["policy_intensity_level"])
high_policy_data = high_policy_data.drop(columns=["policy_intensity_level"])
```

```python
# Save the three policy-level datasets
low_policy_data.to_csv(
    "../Cleaned data/low_policy_data.csv",
    index=False
)

medium_policy_data.to_csv(
    "../Cleaned data/medium_policy_data.csv",
    index=False
)

high_policy_data.to_csv(
    "../Cleaned data/high_policy_data.csv",
    index=False
)
```
